```python
df_model['sex']=df_model['sex'].str.strip()
df_model['smoker']=df_model['smoker'].str.strip()
df_model['region']=df_model['region'].str.strip()
```

```python
scaler= StandardScaler()
```

```python
# Age groups
X_train['age_group'] = pd.cut(
    X_train['age'],
    bins=[0, 18, 40, 60, float('inf')],
    labels=['Children', 'Young', 'Middle-aged', 'Senior']
)

X_test['age_group'] = pd.cut(
    X_test['age'],
    bins=[0, 18, 40, 60, float('inf')],
    labels=['Children', 'Young', 'Middle-aged', 'Senior']
)

# BMI categories
X_train['bmi_category'] = pd.cut(
    X_train['bmi'],
    bins=[0, 18.5, 25, 30, float('inf')],
    labels=['Underweight', 'Normal', 'Overweight', 'Obese']
)

X_test['bmi_category'] = pd.cut(
    X_test['bmi'],
    bins=[0, 18.5, 25, 30, float('inf')],
    labels=['Underweight', 'Normal', 'Overweight', 'Obese']
)
```

```python
X_train=pd.get_dummies(X_train,columns=['sex','region','smoker'],drop_first=True)
X_test=pd.get_dummies(X_test,columns=['sex','region','smoker'],drop_first=True)
```

```python
cat_features = [
    'age', 'bmi', 'children', 'sex_male',
    'region_northwest', 'region_southeast',
    'region_southwest', 'smoker_yes',
    'bmi_category_Underweight', 'bmi_category_Normal',
    'bmi_category_Overweight', 'bmi_category_Obese',
    'age_group_Children', 'age_group_Young',
    'age_group_Middle-aged', 'age_group_Senior'
]
```

```python
cat_features = [col.strip() for col in cat_features]
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

# Bin training target
charges_bin_train = pd.qcut(
    Y_train, q=4, labels=False
)

chi2_results_train = {}

for col in cat_features:
    contingency = pd.crosstab(
        X_train[col], charges_bin_train
    )

    chi2_stat, p_val, _, _ = chi2_contingency(contingency)

    decision = (
        'Reject Null (Keep Feature)'
        if p_val < alpha
        else 'Fail to Reject Null'
    )

    chi2_results_train[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df_train = pd.DataFrame(chi2_results_train).T
chi2_df_train = chi2_df_train.sort_values(by='p_value')

chi2_df_train
```

```python
model.fit(X_train,Y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(Y_test,y_pred)
r2
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1 - ((1- r2) * (n-1) / (n - p - 1))
adjusted_r2
```
